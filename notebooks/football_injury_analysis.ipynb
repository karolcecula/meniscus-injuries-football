```python
# Definition of Meniscus injury 'mensicus...'
player_injuries.loc[
    player_injuries["injury_reason"].str.contains("meniscus", case=False, na=False),
    "injury_reason"
].unique()
```

```python
# table of players with Meniscus injury 
meniscus_df = player_injuries[player_injuries['injury_reason'].isin(['Meniscus damage', 'Meniscus tear', 'Meniscus injury',
       'Tear of the lateral meniscus', 'Meniscus irritation'])].copy()
meniscus_df.shape
```

```python
# change data type
meniscus_df["from_date"] = pd.to_datetime(meniscus_df["from_date"])
meniscus_df["end_date"] = pd.to_datetime(meniscus_df["end_date"])
```

```python
# Change table to proper dates
meniscus_dates = meniscus_df[
    meniscus_df["from_date"].between("2012-07-03", "2026-06-28")
].copy()
```

```python
top5_ids = ["GB1", "ES1", "IT1", "L1", "FR1"]

player_top5 = (
    player_performances[
        player_performances["competition_id"].isin(top5_ids)
    ][["player_id", "season_name", "competition_id"]]
    .drop_duplicates()
)
```

```python
meniscus_top5 = meniscus_dates.merge(
    player_top5,
    on=["player_id", "season_name"],
    how="inner"
)
meniscus_top5.head()
```
